```python
import datetime
import random
```

```python
def chatbot_response(user_input):

    user_input = user_input.lower().strip()

    # Greetings
    if user_input in ["hi", "hello", "hey", "hai"]:
        return random.choice([
            "Hello! How can I help you?",
            "Hi! Nice to meet you.",
            "Hey! What can I do for you?"
        ])

    # Name
    elif "your name" in user_input or "who are you" in user_input:
        return "I am a Rule-Based Chatbot created using Python."

    # How are you
    elif "how are you" in user_input:
        return "I am doing great! Thank you for asking."

    # User name
    elif "my name" in user_input:
        return "Nice to meet you! I am happy to chat with you."

    # Python
    elif "python" in user_input:
        return "Python is a popular programming language used in AI, Data Science, Web Development and Automation."

    # Artificial Intelligence
    elif "ai" in user_input or "artificial intelligence" in user_input:
        return "Artificial Intelligence allows computers to perform tasks that normally require human intelligence."

    # Machine Learning
    elif "machine learning" in user_input:
        return "Machine Learning is a branch of AI where computers learn patterns from data."

    # College
    elif "college" in user_input:
        return "College is a place where students learn technical and professional skills."

    # Study
    elif "study" in user_input or "exam" in user_input:
        return "Create a timetable, study important topics and practice previous questions."

    # Time
    elif "time" in user_input:
        current_time = datetime.datetime.now().strftime("%I:%M %p")
        return "The current time is " + current_time

    # Date
    elif "date" in user_input:
        current_date = datetime.datetime.now().strftime("%d-%m-%Y")
        return "Today's date is " + current_date

    # Joke
    elif "joke" in user_input:
        return random.choice([
            "Why do programmers prefer dark mode? Because light attracts bugs!",
            "Why was the computer cold? Because it left its Windows open!",
            "A programmer's favorite place? The Python!"
        ])

    # Thank you
    elif "thank you" in user_input or "thanks" in user_input:
        return "You're welcome! 😊"

    # Help
    elif "help" in user_input:
        return """You can ask me:
1. What is your name?
2. What is Python?
3. What is AI?
4. What is Machine Learning?
5. What is the time?
6. What is today's date?
7. Tell me a joke
8. Study tips
9. Bye"""

    # Goodbye
    elif user_input in ["bye", "exit", "quit", "goodbye"]:
        return "Goodbye! Thank you for chatting with me."

    # Unknown input
    else:
        return "Sorry, I don't understand that. Please type 'help' to see available questions."
```

```python
print("\n🤖 Bot: Hello! How can I help you?")

while True:

    user_input = input("👤 You: ")

    response = chatbot_response(user_input)

    print("🤖 Bot:", response)

    if user_input.lower().strip() in ["bye", "exit", "quit", "goodbye"]:
        break
```
